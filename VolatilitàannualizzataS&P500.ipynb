```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Download dati S&P 500 disattivando il MultiIndex
data = yf.download("^GSPC", start="2023-01-01", multi_level_index=False)['Close']

# Calcolo rendimenti logaritmici e volatilità
log_returns = np.log(data / data.shift(1)).dropna()
volatility = log_returns.std() * np.sqrt(252)

# Estrazione del valore numerico scalare per la stampa
vol_value = volatility.item() if hasattr(volatility, 'item') else float(volatility)
print(f"Volatilità Annualizzata S&P 500: {vol_value:.2%}")

# Grafico dei rendimenti
log_returns.plot(title="Rendimenti S&P 500", figsize=(10, 4))
plt.show()
```
